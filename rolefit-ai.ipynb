```python
student_skills = ["python", "git", "html", "css"]

job_skills = ["python", "sql", "excel", "data analysis"]

print("My skills:", student_skills)
print("Job needs:", job_skills)
```

```python
matched_skills = []
missing_skills = []

for skill in job_skills:
    if skill in student_skills:
        matched_skills.append(skill)
    else:
        missing_skills.append(skill)

match_score = (len(matched_skills) / len(job_skills)) * 100

print("Matched skills:", matched_skills)
print("Missing skills:", missing_skills)
print("Match score:", match_score, "%")
```

```python
if match_score >= 75:
    result = "Strong match — you should apply."
elif match_score >= 50:
    result = "Partial match — learn the missing skills, then apply."
else:
    result = "Low match — build more skills before applying."

print(result)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

student_profile = """
I know Python, SQL and Git.
I have basic data analysis skills.
I am a Diploma Computer Engineering student.
"""

job_description = """
Junior Data Analyst internship.
Required skills: Python, SQL, Excel, data cleaning,
data visualization and communication.
"""

vectorizer = TfidfVectorizer(stop_words="english")
vectors = vectorizer.fit_transform([student_profile, job_description])

ai_score = cosine_similarity(vectors[0], vectors[1])[0, 0] * 100

print(f"AI similarity score: {ai_score:.1f}%")
```

```python
final_score = (0.6 * match_score) + (0.4 * ai_score)

print(f"Final RoleFit score: {final_score:.1f}%")

if final_score >= 75:
    print("Strong match — you should apply.")
elif final_score >= 50:
    print("Partial match — learn the missing skills, then apply.")
else:
    print("Low match — build more skills before applying.")
```

```python
def get_skill_score(student_skills, job_skills):
    matched = []
    missing = []

    for skill in job_skills:
        if skill in student_skills:
            matched.append(skill)
        else:
            missing.append(skill)

    score = (len(matched) / len(job_skills)) * 100
    return score, matched, missing


web_developer_skills = ["html", "css", "javascript", "git"]

score, matched, missing = get_skill_score(
    student_skills,
    web_developer_skills
)

print(f"Web Developer match: {score:.1f}%")
print("Matched:", matched)
print("Missing:", missing)
```

```python
roles = {
    "Data Analyst": ["python", "sql", "excel", "data analysis"],
    "Web Developer": ["html", "css", "javascript", "git"],
    "QA Tester": ["testing", "python", "sql", "git"]
}

results = []

for role_name, required_skills in roles.items():
    score, matched, missing = get_skill_score(student_skills, required_skills)
    results.append([role_name, score, matched, missing])

results.sort(key=lambda item: item[1], reverse=True)

for role_name, score, matched, missing in results:
    print(f"\n{role_name}: {score:.1f}%")
    print("Matched:", matched)
    print("Missing:", missing)
```

```python
import matplotlib.pyplot as plt

role_names = [item[0] for item in results]
role_scores = [item[1] for item in results]

plt.figure(figsize=(8, 4))
plt.bar(role_names, role_scores, color="royalblue")

plt.ylim(0, 100)
plt.ylabel("Match score (%)")
plt.title("My RoleFit AI Results")

for index, score in enumerate(role_scores):
    plt.text(index, score + 2, f"{score:.1f}%", ha="center")

plt.show()
```
